# SafeSathi — Voice Tone Classifier (scream / distress detection)

Trains a classifier over acoustic features — pitch, loudness, spectral
centroid, pitch jitter — to distinguish `normal_speech`, `fear_distress`,
and `scream`. This replaces the fixed-threshold formula that
`app/services/tone_analysis_service.py` used previously with a genuine
trained model, loaded the same way the motion classifier is.

**Why only 3 classes, not 4 (no "silence" class):** `tone_analysis_service.py`
already gates out near-silent clips *before* any classification happens
(`MIN_RMS_FOR_ANALYSIS` — see that file), returning zero-probability
immediately. So every clip that reaches this classifier is already known
to contain real vocal energy — silence isn't a decision this model needs
to make, and training it on a class it will never actually be asked about
in production would just be padding, not real coverage.

**Why synthetic data, stated plainly:** no labeled dataset of real
screams/distress-speech exists to train on ethically — collecting one
means recording people in actual distress. Each class below is instead
generated from **documented acoustic-science correlates of screaming and
vocal fear** (elevated pitch, loudness, spectral brightness, and — for
fear specifically — pitch instability/jitter from vocal tremor), the same
literature `tone_analysis_service.py`'s previous threshold constants were
already citing. This notebook doesn't invent new claims — it turns
already-cited acoustic reasoning into trainable feature distributions
instead of hand-picked thresholds.


```python
import json
from dataclasses import dataclass
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import cross_val_score, train_test_split

sns.set_theme(style="whitegrid")
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "app").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_DIR = PROJECT_ROOT / "data"
ARTIFACTS_DIR = PROJECT_ROOT / "app" / "ml_artifacts"
DATA_DIR.mkdir(parents=True, exist_ok=True)
ARTIFACTS_DIR.mkdir(parents=True, exist_ok=True)
```

## 1. Configuration

Feature order and class labels must match what
`tone_analysis_service.py` computes per clip and sends to the model at
inference time.


```python
SAMPLES_PER_CLASS = 700

FEATURE_NAMES = ["pitch_hz", "loudness_db", "spectral_centroid_hz", "pitch_jitter"]
CLASS_LABELS = ["normal_speech", "fear_distress", "scream"]
```

## 2. Per-class distributions — and the reasoning behind each

| Class | Reasoning |
|---|---|
| `normal_speech` | Typical conversational F0 (~120–220 Hz), moderate loudness, moderate spectral centroid, **low** pitch jitter — a steady, unstressed voice. |
| `fear_distress` | Elevated pitch from vocal-fold tension under stress, but the defining signature is **high jitter** — the wavering/trembling quality of a frightened voice is a well-established fear correlate distinct from screaming, which is why jitter (not just pitch) matters here. |
| `scream` | Very high pitch and loudness (matching this project's original `SCREAM_PITCH_FLOOR/CEILING_HZ` and `SCREAM_LOUDNESS_*` constants), and the **highest spectral centroid** — the harsh, bright timbre of a scream, from acoustic-science literature on vocal effort and formant behavior at vocal extremes. |

Each profile is a `(mean, std)` pair per feature, sampled from a
(clipped-to-non-negative) normal distribution — same methodology as the
motion classifier notebook.


```python
@dataclass
class ClassProfile:
    pitch_hz: tuple
    loudness_db: tuple
    spectral_centroid_hz: tuple
    pitch_jitter: tuple


CLASS_PROFILES = {
    "normal_speech": ClassProfile(
        pitch_hz=(170, 35), loudness_db=(-22, 5), spectral_centroid_hz=(1300, 300), pitch_jitter=(0.08, 0.03),
    ),
    "fear_distress": ClassProfile(
        pitch_hz=(260, 50), loudness_db=(-16, 5), spectral_centroid_hz=(1800, 350), pitch_jitter=(0.28, 0.08),
    ),
    "scream": ClassProfile(
        pitch_hz=(650, 130), loudness_db=(-8, 4), spectral_centroid_hz=(2900, 500), pitch_jitter=(0.20, 0.07),
    ),
}
```

## 3. Generate readings and save to CSV — the "data collection" step

```python
def _sample_positive(rng, mean_std):
    mean, std = mean_std
    return float(max(0.0, rng.normal(mean, std)))


def generate_readings(samples_per_class: int, seed: int) -> pd.DataFrame:
    rng = np.random.default_rng(seed)
    rows = []
    for label in CLASS_LABELS:
        profile = CLASS_PROFILES[label]
        for _ in range(samples_per_class):
            rows.append({
                "pitch_hz": _sample_positive(rng, profile.pitch_hz),
                "loudness_db": rng.normal(*profile.loudness_db),  # dB is naturally negative, don't clip
                "spectral_centroid_hz": _sample_positive(rng, profile.spectral_centroid_hz),
                "pitch_jitter": _sample_positive(rng, profile.pitch_jitter),
                "label": label,
            })
    return pd.DataFrame(rows)


raw_df = generate_readings(SAMPLES_PER_CLASS, RANDOM_SEED)
csv_path = DATA_DIR / "tone_training_data.csv"
raw_df.to_csv(csv_path, index=False)
print(f"Saved {len(raw_df)} rows to {csv_path}")
```

## 4. Load the dataset from disk

```python
df = pd.read_csv(csv_path)
print(f"Loaded {df.shape[0]} rows, {df.shape[1]} columns from {csv_path}")
df.head()
```

## 5. Clean the data

```python
print("Missing values per column:")
print(df.isnull().sum())

duplicate_count = df.duplicated().sum()
print(f"\nExact duplicate rows: {duplicate_count}")

print("\nClass balance:")
print(df["label"].value_counts())

negative_counts = (df[["pitch_hz", "spectral_centroid_hz", "pitch_jitter"]] < 0).sum()
print(f"\nNegative-value cells on features that must be non-negative (should be 0): {negative_counts.sum()}")

df = df.drop_duplicates().dropna()
print(f"\nRows after cleaning: {len(df)}")
```

### EDA — do the classes actually look separable?

```python
df.groupby("label")[FEATURE_NAMES].mean().loc[CLASS_LABELS]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

sns.boxplot(data=df, x="label", y="pitch_hz", order=CLASS_LABELS, ax=axes[0])
axes[0].set_title("Pitch by class")
axes[0].tick_params(axis="x", rotation=20)

sns.boxplot(data=df, x="label", y="pitch_jitter", order=CLASS_LABELS, ax=axes[1])
axes[1].set_title("Pitch jitter by class\n(the fear-vs-scream signature)")
axes[1].tick_params(axis="x", rotation=20)

sns.boxplot(data=df, x="label", y="spectral_centroid_hz", order=CLASS_LABELS, ax=axes[2])
axes[2].set_title("Spectral centroid by class")
axes[2].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

`fear_distress` should show visibly higher jitter than either
`normal_speech` or `scream`, even though `scream` has the highest pitch —
that's the whole point of including jitter as its own feature rather than
relying on pitch alone.

## 6. Train / test split


```python
X = df[FEATURE_NAMES].to_numpy()
y = df["label"].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)
print(f"Train: {X_train.shape[0]} samples · Test: {X_test.shape[0]} samples")
```

## 7. Train

```python
clf = RandomForestClassifier(
    n_estimators=200, max_depth=6, min_samples_leaf=5,
    random_state=RANDOM_SEED, class_weight="balanced",
)
clf.fit(X_train, y_train)
```

## 8. Evaluate

```python
y_pred = clf.predict(X_test)
report = classification_report(y_test, y_pred, output_dict=True)
print(classification_report(y_test, y_pred))
```

```python
cv_scores = cross_val_score(clf, X, y, cv=5)
print(f"5-fold CV accuracy: {cv_scores.mean():.4f} +/- {cv_scores.std():.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred, labels=CLASS_LABELS)
fig, ax = plt.subplots(figsize=(6, 5.5))
ConfusionMatrixDisplay(cm, display_labels=CLASS_LABELS).plot(ax=ax, cmap="Oranges", colorbar=False)
plt.xticks(rotation=20)
plt.title("Confusion matrix — test set")
plt.tight_layout()
plt.show()
```

```python
importances = sorted(zip(FEATURE_NAMES, clf.feature_importances_), key=lambda p: -p[1])
fig, ax = plt.subplots(figsize=(6.5, 3.5))
names, values = zip(*importances)
sns.barplot(x=list(values), y=list(names), ax=ax, color="#D97757")
ax.set_title("Feature importances")
plt.tight_layout()
plt.show()
for name, importance in importances:
    print(f"  {name}: {importance:.4f}")
```

## 9. Save the model as `.pkl`

`app/core/settings.py` → `tone_classifier_path` points at this exact
file — `tone_analysis_service.py` loads it the same way
`motion_classifier_service.py` loads its model.


```python
import joblib

model_path = ARTIFACTS_DIR / "tone_classifier.pkl"
joblib.dump(clf, model_path)

metadata = {
    "feature_names": FEATURE_NAMES,
    "class_labels": CLASS_LABELS,
    "trained_on": str(csv_path.relative_to(PROJECT_ROOT)),
    "samples_per_class": SAMPLES_PER_CLASS,
    "test_accuracy": report["accuracy"],
    "cv_accuracy_mean": float(cv_scores.mean()),
    "cv_accuracy_std": float(cv_scores.std()),
    "per_class_f1": {label: report[label]["f1-score"] for label in CLASS_LABELS},
}
metadata_path = ARTIFACTS_DIR / "tone_classifier_metadata.json"
metadata_path.write_text(json.dumps(metadata, indent=2))

print(f"Saved model to    {model_path}")
print(f"Saved metadata to {metadata_path}")
```

## 10. How this connects to the rest of SafeSathi

- `tone_analysis_service.py` extracts `pitch_hz`, `loudness_db`,
  `spectral_centroid_hz` (its existing `mean_centroid`), and
  `pitch_jitter` from every voice clip already — those exact features
  now feed this trained model instead of the old fixed-weight formula.
  `scream_probability` and `fear_probability` become this model's
  `predict_proba()` output for the `scream` / `fear_distress` classes,
  keeping the exact same output contract the backend already consumes.
- `data/tone_training_data.csv` and `app/ml_artifacts/tone_classifier.pkl`
  are both committed to git — retraining + `git push` is the whole
  deployment step, same as the motion classifier.
